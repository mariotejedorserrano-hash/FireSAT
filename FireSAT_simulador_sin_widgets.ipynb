```python
# CONFIGURACIÓN — MODIFICA ESTOS VALORES Y VUELVE A EJECUTAR ESTA CELDA Y LA SIGUIENTE
num_satelites = 6
formato = "8U"               # 2U, 4U, 6U, 8U
masa_por_satelite_kg = {"2U": 4, "4U": 8, "6U": 12, "8U": 16}[formato]
num_planos = 2
altura_km = 525
inclinacion_grados = 97.5
tolerancia_inclinacion = 1.0
tolerancia_altura_km = 50
estrategia = "Una campaña por plano"  # "Todos juntos + OTV" o "Dedicado por plano"
origen_transporte = "Spain"  # Spain, France, USA, Other
coste_dispensador_por_sat = 35000
coste_integracion_por_campana = 75000
coste_otv_por_plano_extra = 800000
reserva_porcentaje = 10

shipping={'Spain':{'USA':32000,'French Guiana':23000,'New Zealand':50000,'India':42000,'Japan':48000,'Norway':26000,'United Kingdom':22000,'Brazil':46000,'China':65000,'Other':45000},'France':{'USA':30000,'French Guiana':18000,'New Zealand':48000,'India':40000,'Japan':46000,'Norway':24000,'United Kingdom':20000,'Brazil':45000,'China':65000,'Other':45000},'USA':{'USA':12000,'French Guiana':35000,'New Zealand':47000,'India':44000,'Japan':42000,'Norway':39000,'United Kingdom':35000,'Brazil':38000,'China':65000,'Other':45000},'Other':{'USA':40000,'French Guiana':40000,'New Zealand':45000,'India':45000,'Japan':45000,'Norway':40000,'United Kingdom':38000,'Brazil':42000,'China':65000,'Other':45000}}

def calcular():
    if num_planos > num_satelites or num_planos < 1:
        raise ValueError('Número de planos inválido')
    if formato not in ('2U','4U','6U','8U'):
        raise ValueError('Formato inválido')
    familia='SSO' if 96<=inclinacion_grados<=102 else 'Polar' if inclinacion_grados>=85 else 'Mid-LEO'
    campanas=1 if estrategia=='Todos juntos + OTV' else num_planos
    cantidades=[num_satelites//campanas+(i<num_satelites%campanas) for i in range(campanas)]
    filas=[]
    for _,r in base.iterrows():
        problemas=[]
        if not (r['Min inclination °']-tolerancia_inclinacion<=inclinacion_grados<=r['Max inclination °']+tolerancia_inclinacion): problemas.append('Inclinación')
        if not (r['Min altitude km']-tolerancia_altura_km<=altura_km<=r['Max altitude km']+tolerancia_altura_km): problemas.append('Altitud')
        if familia not in r['Orbit families'].split(','): problemas.append('Familia orbital')
        especulativo=any(x in r['Commercial status'].upper() for x in ['DEVELOPMENT','UNCONFIRMED','SITE TBD','EXPORT RESTRICTIONS','PROGRAM STATUS','SITE / MANIFEST','RFQ'])
        estado='INCOMPATIBLE' if problemas else 'POR CONFIRMAR' if especulativo else 'FILTRO SUPERADO'
        lanzamiento=sum(r['Dedicated USD'] if estrategia=='Dedicado por plano' else r['Rideshare minimum USD']+max(0,c*masa_por_satelite_kg-r['Included kg'])*r['Extra USD/kg'] for c in cantidades)
        dispensadores=num_satelites*coste_dispensador_por_sat
        integracion=campanas*coste_integracion_por_campana
        transporte=campanas*shipping[origen_transporte].get(r['Country'],45000)
        aduanas=campanas*12000
        otv=(num_planos-1)*coste_otv_por_plano_extra if campanas==1 else 0
        subtotal=lanzamiento+dispensadores+integracion+transporte+aduanas+otv
        reserva=subtotal*reserva_porcentaje/100
        filas.append({'Estado':estado,'Proveedor':r['Provider'],'Lanzador':r['Launcher'],'Base':r['Spaceport'],'Campañas':campanas,'Sat/campaña':'+'.join(map(str,cantidades)),'Lanzamiento $':lanzamiento,'Dispensadores $':dispensadores,'Integración $':integracion,'Transporte $':transporte,'Aduanas $':aduanas,'OTV $':otv,'Reserva $':reserva,'TOTAL $':subtotal+reserva,'Coste/sat $':(subtotal+reserva)/num_satelites,'Motivo':', '.join(problemas),'Estado comercial':r['Commercial status']})
    resultados=pd.DataFrame(filas).sort_values('TOTAL $')
    compatibles=resultados[resultados['Estado']!='INCOMPATIBLE']
    print(f'Configuración: {num_satelites} × {formato}, {num_planos} planos, {altura_km} km, {inclinacion_grados}°')
    print(f'{len(compatibles)} opciones superan el filtro preliminar de {len(resultados)} evaluadas')
    print('ATENCIÓN: los filtros no verifican RAAN, LTAN, dispersión, acceso real a rideshare ni factibilidad de OTV.')
    display(compatibles[['Estado','Proveedor','Lanzador','Base','Campañas','TOTAL $','Coste/sat $','Estado comercial']].style.format({'TOTAL $':'${:,.0f}','Coste/sat $':'${:,.0f}'}))
    print('DESGLOSE DE TODAS LAS OPCIONES:')
    display(resultados.style.format({c:'${:,.0f}' for c in resultados.columns if c.endswith('$')}))
    resultados.to_csv('resultados_firesat.csv',index=False,encoding='utf-8-sig')
    print('CSV guardado: resultados_firesat.csv')
    return resultados

resultados_firesat = calcular()
```
